```python
from manim import *

class Intro(Scene):
    def construct(self):
        # Create a circle
        # circle = Circle()
        # circle.set_fill(PINK, opacity=0.5)
        
        # # Create a square
        # square = Square()

        axes = Axes(
            x_range=[-3, 3, 1],   # min, max, step
            y_range=[-3, 3, 1],
            x_length=6,            # width on screen
            y_length=4,            # height on screen
            axis_config={
                "include_tip": True,
                "numbers_to_include": range(-3, 4),
            },
            tips=False,
        )

        axes_labels = axes.get_axis_labels(x_label="x", y_label="f(x)")
        
        # Plot the function
        graph = axes.plot(lambda x: (x-0)**2, color=BLUE, x_range=[-3, 3])

        x0 = 1
        y0 = (x0)**2
        point = axes.c2p(x0, y0)
        
        # Derivative (slope) at x0: f'(x) = 2x, so f'(1) = 2
        slope = 2 * x0
        
        # Tangent line: y = slope * (x - x0) + y0
        # We plot it across a small range around x0 for visibility
        tangent = axes.plot(
            lambda x: slope * (x - x0) + y0,
            color=RED,
            x_range=[x0 - 1.5, x0 + 1.5]
        )

        dot = Dot(point=point)
        
        
        # Optional: add a label on the curve
        graph_label = axes.get_graph_label(
            graph, 
            label="x^2", 
            x_val=2, 
            direction=UP
        )
        
        # Animate them
        self.play(Create(axes))
        self.play(Create(graph), Create(tangent), Create(dot))
        self.play(Write(graph_label))
        self.wait(1)
```

```python
%manim -qm Intro
```

```python
from manim import *

class Derivative2(Scene):
    def construct(self):
        axes = Axes(
            x_range=[-3, 3, 1],  
            y_range=[-3, 3, 1],
            x_length=6,
            y_length=6,
            axis_config={
                "include_tip": True,
                "numbers_to_include": range(-3, 4),
            },
            tips=False,
        )

        axes_labels = axes.get_axis_labels(x_label="x", y_label="f(x)")
        
        def f(x,a=0, w=1, y=0):
            return ((a-(x*w+y))**2)/2

        def d(x,a=0,w=1,y=0):
            return -w*(a-(x*w+y))
        
        # Plot the function
        graph = axes.plot(lambda x: f(x), color=BLUE, x_range=[-3, 3])
        derivative = axes.plot(lambda x: d(x), color=GREEN, x_range=[-3, 3])
        ggraph = axes.plot(lambda x: -d(x), color=PINK, x_range=[-3,3])

        tracker = ValueTracker(-2)
        def tang():
            x0 = tracker.get_value()
            y0 = f(x0)
            x1 = x0+0.001
            y1 = f(x1)
            slope = (y1-y0)/(x1-x0)
            point = axes.c2p(x0, y0)
            point2= axes.c2p(x1, y1)
            return VGroup(axes.plot( lambda x: slope * (x - x0) + y0,
                                color=RED,
                                x_range=[x0 - 1.5, x0 + 1.5]), Dot(point=point))
            
        
        
        # Derivative (slope) at x0: f'(x) = 2x, so f'(1) = 2
        
        # Tangent line: y = slope * (x - x0) + y0
        # We plot it across a small range around x0 for visibility
        tangent = always_redraw(   tang        )

        
        
        # Optional: add a label on the curve
        graph_label = axes.get_graph_label(
            graph, 
            label="(x_{lj}-\\sum_{x_{(l-1)k}}^{N_{(l-1)}}x_{(l-1)k}*w_{(l-1)k})^2", 
            x_val=2, 
            direction=UP
        )
        
        # Animate them
        self.play(Create(axes))
        self.play(Create(graph),Create(ggraph),Create(derivative))
        self.play(Write(graph_label))
        self.play(Create(tangent))
        self.play(tracker.animate.set_value(2), run_time=4)
        self.play(tracker.animate.set_value(-2), run_time=4)
        self.wait(5)

%manim -qm Derivative2
```
