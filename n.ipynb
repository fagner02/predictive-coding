```python
from manim import *

class Intro(Scene):
    def construct(self):
        p1 = MathTex("Mb_{mj}=\\sum_{k}B_{mk}Ma_{kj}").to_corner(UL)
        p2 = MathTex("\\frac{\\partial Mb_{ij}}{\\partial Ma_{mn}}=B_{im} \\cdot \\delta_{jn}").next_to(p1, DOWN).to_edge(LEFT)
        p3= MathTex("\\frac{\\partial P}{\\partial Ma_{mn}} = \\sum_{i,j} \\frac{\\partial P}{\\partial Mb_{ij}} \\cdot \\frac{\\partial Mb_{ij}}{\\partial Ma_{mn}}").next_to(p2, DOWN).to_edge(LEFT)
        p4= MathTex("\\frac{\\partial P}{\\partial Ma_{mn}} = \\sum_{i,j} \\frac{\\partial P}{\\partial Mb_{ij}} \\cdot B_{im} \\cdot \\delta_{jn}").next_to(p3, DOWN).to_edge(LEFT)
        
        p5= MathTex("\\frac{\\partial P}{\\partial Ma_{mn}} = \\sum_{i} \\frac{\\partial P}{\\partial Mb_{in}} \\cdot B_{im}").next_to(p4, RIGHT).to_edge(UP)
        p6=MathTex("\\frac{\\partial P}{\\partial Ma_{mn}}=(B^T)_{mi}\\cdot \\frac{\\partial P}{\\partial Mb_{in}}").next_to(p5, DOWN)
        p7=MathTex("\\frac{\\partial P}{\\partial Ma}=B^T\\cdot \\frac{\\partial P}{\\partial Mb}").next_to(p6, DOWN)

        g2 =VGroup(p1,p2,p3,p4,p5,p6,p7)
        
        self.play(Write(g2))
        self.wait(1)
%manim -qm Intro
```

```python
from manim import *

class Intro2(Scene):
    def construct(self):
        p1 = MathTex("Ma_{mj}=\\sum_{k}X_{mk}A_{kj}").to_corner(UL)
        p2 = MathTex("\\frac{\\partial Ma_{ij}}{\\partial X_{mn}}= \\delta_{im} \\cdot A_{nj}").next_to(p1, DOWN).to_edge(LEFT)
        p3= MathTex("\\frac{\\partial Mb}{\\partial X_{mn}} = \\sum_{i,j} \\frac{\\partial Mb}{\\partial Ma_{ij}} \\cdot \\frac{\\partial Ma_{ij}}{\\partial X_{mn}}").next_to(p2, DOWN).to_edge(LEFT)
        p4= MathTex("\\frac{\\partial Mb}{\\partial X_{mn}} = \\sum_{i,j} \\frac{\\partial Mb}{\\partial Ma_{ij}} \\cdot \\delta_{im} \\cdot A_{nj}").next_to(p3, DOWN).to_edge(LEFT)
        
        p5= MathTex("\\frac{\\partial Mb}{\\partial X_{mn}} = \\sum_{j} \\frac{\\partial Mb}{\\partial Ma_{mj}} \\cdot A_{nj}").next_to(p4, RIGHT).to_edge(UP)
        p6=MathTex("\\frac{\\partial Mb}{\\partial X_{mn}}= \\sum_{j} \\frac{\\partial Mb}{\\partial Ma_{mj}} \\cdot (A^T)_{jn}").next_to(p5, DOWN)
        p7=MathTex("\\frac{\\partial Mb}{\\partial X}=\\frac{\\partial Mb}{\\partial X} \\cdot A^T").next_to(p6, DOWN)

        g2 =VGroup(p1,p2,p3,p4,p5,p6,p7)
        
        self.play(Write(g2))
        self.wait(1)
%manim -qm Intro2
```

```python
from manim import *

class Intro3(Scene):
    def construct(self):
        e = MathTex("e=A_l-P_l").to_corner(UR)
        p = MathTex("P_l=\\sum_{X_{l-1,i}}^{N_{l-1,i}}Mb \\cdot W_{l,i}").next_to(e, DOWN).to_edge(LEFT)
        b = MathTex("Mb=B \\times Ma").next_to(p, DOWN).to_edge(LEFT)
        a = MathTex("Ma=X_{l-1,i} \\times A").next_to(b, DOWN).to_edge(LEFT)
        g1 = VGroup(e,p,b,a)

        self.play(Write(g1))
        self.wait(1)
%manim -qm Intro3
```

```python
from manim import *
import numpy as np

class SigmoidSmoothLines(Scene):
    def construct(self):
        axes = Axes(
            x_range=[-3, 8, 1],  
            y_range=[-2, 2, 1],
            x_length=12,
            y_length=6,
            axis_config={
                "include_tip": True,
                "numbers_to_include": range(-3, 4),
            },
            tips=False,
        )
        axes_labels = axes.get_axis_labels(x_label="x", y_label="f(x)")
        w11=ValueTracker(2)
        w12=ValueTracker(-3)
        w21=ValueTracker(2)
        w22=ValueTracker(-7)
        wo1=1
        wo2=-1

        def sigmoid(x):
            return 1/(1+np.exp(-x))
        def h1():
            return axes.plot(lambda x: sigmoid(x*w11.get_value()+w12.get_value()), color=BLUE)
        
        def h2():
            return axes.plot(lambda x: sigmoid(x*w21.get_value()+w22.get_value()), color=GREEN)

        def pf(x):
            a = sigmoid(x*w11.get_value()+w12.get_value())
            b = sigmoid(x*w21.get_value()+w22.get_value())
            return wo1*a+wo2*b
        def p():
            return axes.plot(lambda x: pf(x), color=ORANGE)

        def labelsf():
            p3 = MathTex("p=w_1^o\\cdot \\sigma(h_1) + w_2^o\\cdot \\sigma(h_2)").to_corner(DR)
            p2 = MathTex("h_2=w_2\\cdot x+b_2").next_to(p3, UP).to_edge(RIGHT)
            p1 = MathTex("h_1=w_1\\cdot x+b_1").next_to(p2, UP).to_edge(RIGHT)
                                                        
            p4 = MathTex(f"w_1={w11.get_value():.2f},b_1={w12.get_value():.2f}").to_corner(UR)
            p5 = MathTex(f"w_2={w21.get_value():.2f},b_2={w22.get_value():.2f}").next_to(p4, DOWN).to_edge(RIGHT)
            p6 = MathTex(f"w^o_1={wo1:.2f},w^o_2={wo2:.2f}").next_to(p5, DOWN).to_edge(RIGHT)
            return VGroup(p1, p2, p3,p4,p5,p6)
        labels = always_redraw(labelsf)
        graphh1= always_redraw(h1)
        graphh2= always_redraw(h2)
        graphp= always_redraw(p)
        
        # Animate them
        self.add(axes, axes_labels)
        self.play(Create(graphh1), Create(graphh2), Create(graphp))
        self.play(Create(labels))
        self.play(w12.animate.set_value(2),run_time=4)
        self.play(w12.animate.set_value(-14),run_time=4)
        self.play(w12.animate.set_value(-3),run_time=4)

        self.play(w22.animate.set_value(2),run_time=4)
        self.play(w22.animate.set_value(-17),run_time=4)
        self.play(w22.animate.set_value(-7),run_time=4)

        self.play(w11.animate.set_value(-6), run_time=5)
        self.play(w11.animate.set_value(6), run_time=5)
        self.play(w11.animate.set_value(2))

        self.play(w21.animate.set_value(-6), run_time=5)
        self.play(w21.animate.set_value(6), run_time=5)
        self.play(w21.animate.set_value(2))
        
        self.wait(10)

%manim -qm SigmoidSmoothLines
```

```python
from manim import *

class Derivative2(Scene):
    def construct(self):
        axes = Axes(
            x_range=[-3, 3, 1],  
            y_range=[-3, 3, 1],
            x_length=6,
            y_length=6,
            axis_config={
                "include_tip": True,
                "numbers_to_include": range(-3, 4),
            },
            tips=False,
        )

        axes_labels = axes.get_axis_labels(x_label="x", y_label="f(x)")
        
        def f(x,a=0, w=1, y=0):
            return ((a-(x*w+y))**2)/2

        def d(x,a=0,w=1,y=0):
            return -w*(a-(x*w+y))
        
        # Plot the function
        graph = axes.plot(lambda x: f(x), color=BLUE, x_range=[-3, 3])
        derivative = axes.plot(lambda x: d(x), color=GREEN, x_range=[-3, 3])
        ggraph = axes.plot(lambda x: -d(x), color=PINK, x_range=[-3,3])

        tracker = ValueTracker(-2)
        def tang():
            x0 = tracker.get_value()
            y0 = f(x0)
            x1 = x0+0.001
            y1 = f(x1)
            slope = (y1-y0)/(x1-x0)
            point = axes.c2p(x0, y0)
            point2= axes.c2p(x1, y1)
            return VGroup(axes.plot( lambda x: slope * (x - x0) + y0,
                                color=RED,
                                x_range=[x0 - 1.5, x0 + 1.5]), Dot(point=point))
            
        
        
        # Derivative (slope) at x0: f'(x) = 2x, so f'(1) = 2
        
        # Tangent line: y = slope * (x - x0) + y0
        # We plot it across a small range around x0 for visibility
        tangent = always_redraw(   tang        )

        
        
        # Optional: add a label on the curve
        graph_label = axes.get_graph_label(
            graph, 
            label="(x_{lj}-\\sum_{x_{(l-1)k}}^{N_{(l-1)}}x_{(l-1)k}*w_{(l-1)k})^2", 
            x_val=2, 
            direction=UP
        )
        
        # Animate them
        self.play(Create(axes))
        self.play(Create(graph),Create(ggraph),Create(derivative))
        self.play(Write(graph_label))
        self.play(Create(tangent))
        self.play(tracker.animate.set_value(2), run_time=4)
        self.play(tracker.animate.set_value(-2), run_time=4)
        self.wait(5)

%manim -qm Derivative2
```

```python
from manim import *
import numpy as np

class NetworkOutput(ThreeDScene):
    def construct(self):
        axes = ThreeDAxes(
            x_range=[-1, 1, 1],  
            y_range=[-1, 1, 1],
            z_range=[-1, 1, 1],
            x_length=4,
            y_length=4,
            z_length=4,
            tips=False,
        )
        x_label = axes.get_x_axis_label("X")
        y_label = axes.get_y_axis_label("Y")
        z_label = axes.get_z_axis_label("Z")

        def sigmoid(x):
            return 1/(1+np.exp(-x))
        w11=-3.81343
        w12=-3.58818
        w13=0.753555
        w21=-0.932131
        w22=-0.773737
        w23=-0.171377
        wo1=-8.68617
        wo2=11.6513
        wo3=-1.40196
        surface = Surface(
            lambda u, v: np.array([u,v, sigmoid(u*wo1+v*wo2+wo3)]),
            u_range=[-1, 1],
            v_range=[-1, 1],
            resolution=(32, 32), # Keep this relatively low (e.g., 32) or rendering will take forever
            checkerboard_colors=[BLUE_D, BLUE_E],
            fill_opacity=0.8
        )
        self.set_camera_orientation(phi=70 * DEGREES, theta=-45 * DEGREES)
        
        
        # Animate them
        self.add(axes, x_label, y_label, z_label)
        self.play(Create(surface))
        self.begin_ambient_camera_rotation(rate=0.15)
        self.wait(10)

%manim -qm NetworkOutput
```

```python
from manim import *
from scipy import stats as st
import colorsys as csys
import numpy as np

class Fourier(Scene):
    def construct(self):
        axes = Axes(
            x_range=[-3, 8, PI*0.25],  
            y_range=[-1, 1, 0.5],
            y_length=2,
            tips=False
        ).to_edge(UP+LEFT)

        axesc = Axes(
            x_range=[-1,1,1],
            y_range=[-1,1,1],
            x_length=4,
            y_length=4,
            tips=False
        ).next_to(axes, DOWN).to_edge(LEFT)

        freqaxesa = Axes(
            x_range=[0,2*PI,0.5],
            y_range=[0,1,1],
            x_length=8,
            y_length=1,
            tips=False
        ).next_to(axesc.get_corner(UR), DOWN*2+RIGHT*2)
        freqaxesb = Axes(
            x_range=[0,2*PI,0.5],
            y_range=[0,1,1],
            x_length=8,
            y_length=1,
            tips=False
        ).next_to(freqaxesa, DOWN)

        def f(x):
            return (np.sin(x*2)+np.sin(x*3))
            
        c1 = np.array([*csys.hls_to_rgb(h=0.8,s=0.7,l=0.6,),1])
        c2 = np.array([*csys.hls_to_rgb(h=0.1,s=0.8,l=0.6,),1])
        graph = axes.plot(lambda x: f(x), color=c1)
        def circlef(x):
            return np.array([np.cos(x),np.sin(x)])*0.5

        speedTracker = ValueTracker(0)
        def curvef(x):
            h = f(x)
            speed=speedTracker.get_value()
            a = np.sin(x*speed)*h
            b = np.cos(x*speed)*h
            return np.array([a,b])
        
        circle = axesc.plot_parametric_curve(
            circlef,
            t_range=[0,2*PI,0.1],
            color=GRAY
        )
        def curve2(): 
            return axesc.plot_parametric_curve(
                curvef,
                t_range=[0, 4*PI, 0.1],
                color=c2
            )
        curve = always_redraw(curve2)
        freqarr=np.array([[0,0,0]])
        def frequencyf():
            nonlocal freqarr
            speed=speedTracker.get_value()
            def v(x):
                return np.sin(x*speed)*f(x)
            def h(x):
                return np.cos(x*speed)*f(x)
            x = np.linspace(-PI, 4*PI, 500)
            a = np.vectorize(h)(x)
            b = np.vectorize(v)(x)
            
            freqarr = np.append(freqarr, [[speed, a.mean(),b.mean()]], axis=0)
            x = freqarr[:,2]
            y = freqarr[:,1]
            speeds = freqarr[:,0]
            size = freqarr.shape[0]
            m = (axesc.get_corner(UR))
            gh = freqaxesa.plot_line_graph(x_values=speeds,y_values=x,add_vertex_dots=False)
            gv = freqaxesb.plot_line_graph(x_values=speeds,y_values=y,add_vertex_dots=False)
            return VGroup(
                gh,gv,
                Dot(point=axesc.c2p(b.mean(), a.mean()))
            )
        
        frequency = always_redraw(frequencyf)
        # Animate them
        self.add(axes, axesc, freqaxesa, freqaxesb)
        self.play(Create(graph),Create(circle))
        self.play(Create(curve), Create(frequency))
        
        self.play(speedTracker.animate.set_value(2), run_time=20)
        self.play(speedTracker.animate.set_value(3), run_time=10)
        self.play(speedTracker.animate.set_value(PI*2), run_time=20)
        self.wait(10)

%manim -qm Fourier
```

```python
from manim import *
from scipy import stats as st
import colorsys as csys
import numpy as np

class Bound(Scene):
    def construct(self):
        axes = Axes(
            x_range=[0, 4, 1],  
            y_range=[-1, 2, 0.5],
            y_length=4,
            tips=False
        ).to_edge(UP+LEFT)
        xi = 0.5
        wi = 1
        lb = 1
        ub = 2
        p = 0
        def f(x):
            xx = x * wi
            xx = 0 if xx < x else (0 if xx > ub else xx)
            e = p - xx
            return e
        def fd(x):
            x0 = x
            y0 = f(x0)
            x1 = x0+0.1
            y1 = f(x1)
            slope = (y1-y0)/(x1-x0)
            return slope

        graph = axes.plot(lambda x: f(x), color=PINK)
        der = axes.plot(lambda x: fd(x), color=BLUE)
        # Animate them
        self.add(axes)
        self.play(Create(graph), Create(der))
        
        self.wait(10)

%manim -qm Bound
```
